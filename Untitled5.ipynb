```python
data = df[['Production', 'hour', 'month', 'dayofweek']].copy()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(data)
```

```python
import numpy as np

def create_dataset(dataset, time_step=24):
    X, Y = [], []
    for i in range(len(dataset) - time_step - 1):
        X.append(dataset[i:(i + time_step)])
        Y.append(dataset[i + time_step][0])  # predict Production
    return np.array(X), np.array(Y)

time_step = 24
X, y = create_dataset(scaled_data, time_step)

print("X shape:", X.shape)
```

```python
train_size = int(len(X) * 0.8)

X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM

# reshape already correct (samples, time_steps, features)

model_lstm = Sequential()

model_lstm.add(LSTM(64, return_sequences=True, input_shape=(time_step, X.shape[2])))
model_lstm.add(LSTM(32))
model_lstm.add(Dense(1))

model_lstm.compile(loss='mean_squared_error', optimizer='adam')

model_lstm.fit(X_train, y_train, epochs=20, batch_size=32, validation_data=(X_test, y_test))
```

```python
lstm_pred = model_lstm.predict(X_test)
```

```python
X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=100)

rf_model.fit(X_train_flat, y_train)

rf_pred = rf_model.predict(X_test_flat)
```

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

lr_model.fit(X_train_flat, y_train)

lr_pred = lr_model.predict(X_test_flat)
```

```python
def inverse_transform(pred):
    temp = np.zeros((len(pred), scaled_data.shape[1]))
    temp[:,0] = pred
    return scaler.inverse_transform(temp)[:,0]

y_test_actual = inverse_transform(y_test)

lstm_pred_actual = inverse_transform(lstm_pred.flatten())
rf_pred_actual = inverse_transform(rf_pred)
lr_pred_actual = inverse_transform(lr_pred)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

def evaluate(y_true, y_pred, name):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    
    print(f"{name} RMSE:", rmse)
    print(f"{name} R2:", r2)
    print("-"*30)

evaluate(y_test_actual, lstm_pred_actual, "LSTM")
evaluate(y_test_actual, rf_pred_actual, "Random Forest")
evaluate(y_test_actual, lr_pred_actual, "Linear Regression")
```
